# torch.nn.Module

O torch.nn.Module é a classe base para todos os modelos e camadas de redes neurais no PyTorch.

Ele serve para estruturar sua rede e automatizar o gerenciamento de componentes. Suas principais funções são:

- Gerenciamento de Parâmetros: Rastreia automaticamente todos os pesos e vieses (self.parameters()) das camadas internas para o otimizador.
- Organização em Submódulos: Permite empilhar e aninhar camadas (como nn.Linear, nn.Conv2d) e tratá-las como um único bloco.
- Fluxo de Dados (forward): Exige a implementação do método forward(), que define como a entrada passa pelas camadas do modelo.
- Controle de Estado e Dispositivo: Facilita mover o modelo inteiro para GPU/CPU (.to('cuda')), alternar entre treino e avaliação (.train(), .eval()), e salvar/carregar pesos (.state_dict()).

```python
import torch.nn as nn

class MeuModelo(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(10, 2)  # Módulo interno com parâmetros rastreados

    def forward(self, x):
        return self.linear(x)          # Execução do modelo
```
